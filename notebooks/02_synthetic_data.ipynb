```python
import numpy as np
import pandas as pd
np.random.seed(42)

n=10000

age=np.random.randint(21,65,size=n)
income=np.random.lognormal(mean=10.5,sigma=0.5,size=n)
debt_to_income=np.clip(
    np.random.normal(0.35,0.15,size=n),
    0.05,
    0.9
)

credit_score=np.clip(
    np.random.normal(650,70,size=n),
    300,
    850
)

loan_amount=np.clip(
    np.random.normal(0.3*income,0.2*income),
    5000,
    500000
)
df=pd.DataFrame({
    "age":age,
    "income": income,
    "debt_to_income": debt_to_income,
    "credit_score": credit_score,
    "loan_amount": loan_amount
})

df.head()
```

```python
risk_score = (
    0.4 * debt_to_income +
    0.3 * (loan_amount / income) -
    0.002 * credit_score +
    np.random.normal(0, 0.05, size=n)
)
```

```python
threshold = np.percentile(risk_score, 60)
approved = (risk_score < threshold).astype(int)

df["approved"] = approved
```

```python
df.describe()
df["approved"].value_counts(normalize=True)
```

#Assumptions

1. Age matters because it is important to determine earning capacity, number of years of earning, and age of the credit facility. Income matters because it allows us to give an idea of what will be the debt servicing capability and also the disposable income. The debt_to_income score will then give us an idea of how much of the debt can be recovered in case of a default. The credit_score is an indicator of their past credit performance and how they were able to service past debt.

2. Noise is necessary because most datasets have:
- Incomplete information
- Wrongly filled information
- Unnecessary information
- Irrelevant information
- Inaccurate or fraudulent information

3. Imbalance is realistic because data sets usually come from a certain part of a geography, or a polity, or a certain economic group, and therefore will have more people or more individuals of a specific criteria. For example, urban centers might have more salaried employees, as opposed to rural geography data sets where there might be more self-employed individuals.
