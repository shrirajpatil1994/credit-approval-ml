# Credit Approval – Logistic Regression

This notebook demonstrates a simple, end-to-end machine learning workflow:
- data preparation
- train/test split
- model training
- evaluation

This is a toy example focused on methodology, not performance.

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.DataFrame({
    "age": [25, 30, 35, 40, 45],
    "income": [50000, 60000, 70000, 80000, 90000],
    "approved": [0, 0, 1, 1, 1]
})

df
```

```python
X = df[["age", "income"]]
y = df["approved"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.4,
    random_state=42
)
```

```python
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
accuracy
```

#Given the extremely small dataset, this accuracy value is not statistically meaningful.
It is shown only to demonstrate correct evaluation procedure.

### Interpretation

This notebook demonstrates:
- correct data separation
- avoidance of training-test leakage
- use of a baseline model

In real-world credit modeling, additional steps would be required:
- larger datasets
- feature engineering
- cross-validation
- fairness and bias checks
